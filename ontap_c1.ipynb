# câu 1: so sánh lập trình cấu trúc và lập trình hướng đối tượng 

## *khái niệm*
### _*lập trình cấu trúc*_ : 
- là cấu trúc chia chương trình thành hàm module, mỗi chương trình đảm nhiệm một chức năng rõ ràng 
- tập trung vào thuật toán và các luồng điều khiển
- dữ liệu được truyền vào hàm để dễ xử lí

### _*lập trình hướng đối tượng*_: 
- mô hình hóa một thực thể trong thế giới thực, có thuộc tính(attribute), và hành vi(method) riêng 
- có thể tái sử dụng các đối tượng và mở rộng hơn 


## *đặc trung*
### _*lập trình cấu trúc*_ có 3 cấu trúc cơ bản: 
- phân chia trương chình thành các khối: chương trình được chia thanhd các hàm hoặc thủ tục nhỏ, giúp dẽ dang f quản lí và bảo trì 
- sử dụng cấu trúc điều kiểu: 
    - vòng lặp (for, while)
    - điều kiện 
- tính rõ ràng: code được viết rõ ràng, dễ bảo trì, dễ hiểu 

### _*lập trình hướng đối tượng có 4 đặc trưng*_
- đóng gói: che giấu dữ liệu, bảo vệ dữ liệu, đảm bảo tính nhất quán và an toán cho dữ liệu 
- tính kế thừa: là kế thừa từ một lớp khác, giúp tái sử dụng mã nguồn, giảm sự lặp lại của các mã nguồn đồng thời tăng tính linh hoạt 
- tính đa hình: nhiều lớp có những phương thức giống nhau nhưng có thể thực thi theo cách khác nhau
- tính trừu tượng: tách các đặc tính quan trong của đối tượng ra khỏi các chi tiết cụ thể

# câu 3: tính đóng gói của lập trình hướng đối tượng có đặc trưng gì 
# *khái niệm*
Là kỹ thuật giúp che dấu thông tin của đối tượng với môi trường bên ngoài. Tính chất này giúp đảm bảo tính toàn vẹn và bảo mật của đối tượng. Sử dụng OOP trong Python, có thể hạn chế quyền truy cập vào trạng thái bên trong của đối tượng. Điều này ngăn chặn dữ liệu bị sửa đổi trưc tiếp,được gọi là đóng gói (encapsulation).

# *phạm vi sử dụng* 
- Private: không thể truy cập thành viên của lớp khi đứng ngoài lớp. Trong Python, thành viên dạng private sẽ có tiền tố là 2 dấu gạch dưới trước tên thành viên '__'.
- Protected: chỉ truy cập được thành viên của lớp khi đứng ở trong lớp đó hoặc lớp con
của nó. Trong Python, thành viên dạng protected sẽ có tiền tố là một dấu gạch dưới '_' trước tên thành viên.
- Public: mặc định (khai báo như tên thông thường) thì thành viên ở chế độ public, tức là có thể truy xuất vào các thành viên khi đứng ở bất kì đâu (trong lớp, ngoài lớp ... )

    - Đa hình cho phép định nghĩa ra các phương thức trong lớp con mà có cùng tên với phương thức trong lớp cha.

    - Việc ghi đè đặc biệt hữu ích khi phương thức kế thừa từ lớp cha chưa thực sự phù hợp với lớp con.

- method overriding 
    - Công việc cài đặt lại một phương thức được kế thừa từ lớp cha tại lớp con gọi là ghi đè phương thức (Method Overriding).

# câu 6:  khái niệm lớp trừu tượng 
- Nó cho phép người dùng tạo các lớp trừu tượng, định nghĩa hành vi chung mà các lớp con có thể thừa hưởng và triển khai một cách cụ thể. Lớp trừu tượng (abstract class) trong OOP được coi như là bản thiết kế hay bộ khungđể các class khác tuân theo. Một lớp có một hoặc nhiều phương thức trừu tượng gọi là lớp trừu tượng (abstract class).

```python
class hcn:
    def __init__(self, chieu_dai:int, chieu_rong:int):
        self.chieu_dai= chieu_dai
        self.chieu_rong = chieu_rong

    def chu_vi_hcn(self):
        #print("chu vi hình chữ nhật là:{}".format((self.chieu_dai + self.chieu_rong) * 2))
        return (self.chieu_dai + self.chieu_rong) * 2

    def dien_tich_hcn(self):
        #print("dien tich hình chữ nhật là:{}".format(self.chieu_dai * self.chieu_rong))
        return self.chieu_dai * self.chieu_rong

# chieu_dai = float(input("nhap chieu dai:"))
# chieu_rong = float(input("nhap chieu rong:"))
hinh_chu_nhat= hcn(10, 5)
hinh_chu_nhat.chu_vi_hcn()
hinh_chu_nhat.dien_tich_hcn()
```

```python
class hinh_chu_nhat:
    def __init__(self, chieu_dai:int, chieu_rong:float):
        self.chieu_dai= chieu_dai
        self.chieu_rong = chieu_rong

    def tinh_chu_vi_hinh_chu_nhat(self):
        print("chu vi hình chữ nhật là:{}".format((self.chieu_dai + self.chieu_rong) * 2))

    def tinh_dien_tich_hinh_chu_nhat(self):
        print("dien tich hình chữ nhật là:{}".format(self.chieu_dai * self.chieu_rong))

# chieu_dai = float(input("nhap chieu dai:"))
# chieu_rong = float(input("nhap chieu rong:"))
hcn = hinh_chu_nhat(10, 5)
hcn.tinh_chu_vi_hinh_chu_nhat()
```

```python
class quan_li_diem_thi:
    def __init__(self, ten_sinh_vien:str, diem_toan:float, diem_ly:float, diem_hoa:float):
        self.ten_sinh_vien = ten_sinh_vien
        self.diem_toan = diem_toan
        self.diem_ly = diem_ly
        self.diem_hoa = diem_hoa

    def nhap_thong_tin_sinh_vien(self):
        print("Tên sinh viên:", self.ten_sinh_vien)
        print("Điểm Toán:", self.diem_toan)
        print("Điểm Lý:", self.diem_ly)
        print("Điểm Hóa:", self.diem_hoa)

    def tinh_tong_diem_cua_sinh_vien(self):
        tong = self.diem_toan + self.diem_ly + self.diem_hoa
        print("Tổng điểm của sinh viên là:", tong)
        if tong >= 20:
            print(f"Sinh viên {self.ten_sinh_vien} trúng tuyển.")
        else:
            print(f"Sinh viên {self.ten_sinh_vien} không trúng tuyển.")
        return tong

sinh_vien_1= quan_li_diem_thi("Nguyen Van A", 7.5, 6.0, 8.0)
sinh_vien_1.nhap_thong_tin_sinh_vien()
sinh_vien_1.tinh_tong_diem_cua_sinh_vien()
print("===============================")

sinh_vien_2 = quan_li_diem_thi("Le Thi B", 5.0, 4.5, 6.0)
sinh_vien_2.nhap_thong_tin_sinh_vien()
sinh_vien_2.tinh_tong_diem_cua_sinh_vien()

```

```python
class lop_phan_so_PS:
    def __init__(self,tu:int, mau:int):
        self.tu=tu
        self.mau=mau

    def in_phan_so(self):
        print(f"Phân số là: {self.tu}/{self.mau}")

    def gia_tri_phan_so(self):
        gia_tri = self.tu / self.mau
        print(f"Giá trị phân số là: {gia_tri}")
        return gia_tri
    
    def kiem_tra_phan_so(self):
        if self.tu==0:
            print("Phân số bằng 0")
        elif self.tu*self.mau>0:
            print("Phân số dương")
        else:
            print("Phân số âm")
    
phan_so_1= lop_phan_so_PS(3,4)
phan_so_1.in_phan_so()
phan_so_1.gia_tri_phan_so()
phan_so_1.kiem_tra_phan_so()
```

Ngăn xếp (Stack)
- Ngăn xếp (Stack) là một cấu trúc dữ liệu trừu tượng 
- hoạt động theo nguyên tắc LIFO (Last In First Out) - vào sau ra trước.
#### **giống như chồng đĩa, sách: đặt vào sau nhưng lấy ra trước** ####
#### **chỉ được lấy từ phần top(đinh) của ngăn xếp** ####
đặc điểm:
- Chỉ có thể thêm/xóa phần tử ở một đầu (đỉnh stack)
- Phần tử được thêm vào sau cùng sẽ được lấy ra đầu tiên
- Các thao tác cơ bản: push (thêm ở đỉnh), pop (lấy ra phần tử ở đỉnh), peek (xem phần tử đỉnh), size(số phần tử trong ngăn xếp)
- Ví dụ triển khai đơn giản bằng Python:
- Ứng dụng:
    - Quản lý lời gọi hàm trong chương trình
    - Kiểm tra cặp dấu ngoặc hợp lệ
    - Chuyển đổi biểu thức trung tố sang hậu tố
    - Quay lui (backtracking)
    - Undo/Redo trong các ứng dụng

```python
class Stack:
    def __init__(self):
        self.items = []   # danh sách lưu các phần tử trong stack

    # Thêm phần tử vào đỉnh ngăn xếp
    def push(self, item):
        self.items.append(item)

    # Lấy và xóa phần tử ở đỉnh ngăn xếp
    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        else:
            print("Ngăn xếp rỗng!")

    # Xem phần tử ở đỉnh (nhưng không xóa)
    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        else:
            print("Ngăn xếp rỗng!")

    # Kiểm tra rỗng
    def is_empty(self):
        return len(self.items) == 0

    # Đếm số phần tử
    def size(self):
        return len(self.items)


if __name__ == "__main__":
    stack = Stack()
    stack.push(10)
    stack.push(20)
    stack.push(30)
    print("Kích thước ngăn xếp:", stack.size())  # Output: 3
    print("Phần tử ở đỉnh ngăn xếp:", stack.peek())  # Output: 30
    print("Lấy phần tử khỏi ngăn xếp:", stack.pop())  # Output: 30
    print("Kích thước ngăn xếp sau khi pop:", stack.size())  # Output: 2

    stack.push(40)
    print("Kích thước ngăn xếp sau khi thêm 40:", stack.size())  # Output: 3
    print("Phần tử ở đỉnh ngăn xếp:", stack.peek())  # Output: 40
    print("Kích thước ngăn xếp sau khi pop:", stack.size())  # Output: 2
```
